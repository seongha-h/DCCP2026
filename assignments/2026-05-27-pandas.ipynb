```python
import numpy as np
```

```python
df=pd.DataFrame({
    '이름':['강은수','조재영','이한솔','뽀로로'],
    '국어':[88,76,np.nan,77],
    '영어':[89,np.nan, 66, 78]

})
```

```python
df.isna().sum()
df.isna().sum().sum()
```

```python
df.dropna()
```

```python
df.drop(subset=['국어'])
```

```python
df.fillna(0) #모든 결측치를 0으로 채움
```

```python
df[['국어','영어']].mean()
```

```python
def grade(score):
    if score>90:
        return 'A'
    if score>80:
        return 'B'
    if score>70:
        return 'C'
    return 'F'
df['국어등급']=df['국어'].apply(grade)
```

```python
df.groupby('학과')['국어'].mean()
```
