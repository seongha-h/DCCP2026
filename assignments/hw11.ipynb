```python
import os
from dotenv import load_dotenv
load_dotenv()
KEY: str = os.environ["KOREAN_DICT_KEY"]
print(KEY[:4] + "***") # 키 전체가 출력되지 않도록 앞 4자만 확인
```

```python
def search_word(q: str, num: int = 10, start: int = 1) -> dict:
    params={
        "key":KEY, "q":q, "req_type":"json",
        "num":num, "start":start,"type1":"word,"
        }
    r=requests.get(
        "https://opendict.korean.go.kr/api/search",
        params=params, timeout=10,
    )
    r.raise_for_status()
    return r.json()
#수업 슬라이드 그대로 활용
```

```python
import json
import requests
data=search_word('김치')
print(json.dumps(data, ensure_ascii=False, indent=2)[:400])

#수업 슬라이드 그대로 활용
```

```python
#ensure_ascii=False를 빼면 출력 값이 한글이 아닌 유니코드 인코딩 값으로 나온다
```

```python
items = data["channel"]["item"]

print(f"총 {data['channel']['total']}건, 이 페이지 {len(items)}건")

for it in items[:5]:
    word = it["word"]
    pos = it.get("pos", "품사 없음")

    senses = it.get("sense", [])
    if senses:
        definition = senses[0].get("definition", "")
    else:
        definition = ""

    print(f"{word} ({pos}) -> {definition[:40]}")
```

```python
#API 응답 data에서 검색 결과 목록(item)을 꺼낸 뒤, 전체 결과 수와 현재 페이지에 받은 항목 수를 출력한다.
#그다음 첫 5개 항목에 대해 표제어(word), 품사(pos), 뜻풀이(definition)를 꺼내 한 줄씩 출력한다.
#sense나 pos가 없을 수도 있으므로 get()과 if senses:를 사용해 오류가 나지 않도록 처리했다.
```

```python
words: list[str] = [
"김치", "라면", "만두", "김밥",
"국수", "떡볶이", "불고기”", "비빔밥",
]
```

```python
import time
```

```python

all_items = []

for q in words:
    data = search_word(q)

    total = data["channel"]["total"]
    print(f"{q}: {total}건")
    items = data["channel"].get("item", [])
    all_items.extend(items)

    time.sleep(0.3)
```

```python
from collections import Counter

pos_list = []

for item in all_items:
    pos = item.get("pos")

    if not pos:
        senses = item.get("sense", [])
        if senses:
            pos = senses[0].get("pos")

    if not pos:
        pos = "(미상)"

    pos_list.append(pos)

counter = Counter(pos_list)

print("품사 빈도 상위 3개")
for pos, count in counter.most_common(3):
    print(f"{pos}: {count}개")
```

```python
#이 코드는 all_items에 모아둔 검색 결과들에서 품사 정보를 하나씩 꺼내 pos_list에 저장한 뒤, Counter로 품사별 빈도를 센다.
#품사가 item["pos"]에 없으면 sense[0]["pos"]에서 한 번 더 찾고, 그래도 없으면 "(미상)"으로 처리한다.
#마지막으로 most_common(3)을 사용해 가장 많이 나온 품사 3개와 그 개수를 출력한다.
```
