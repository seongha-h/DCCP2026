```python
import numpy as np
import pandas as pd
works = pd.DataFrame({
'작가': ['김유정', '김유정', '김유정',
'현진건', '현진건',
'이상', '이상',
'나도향', '나도향',
'채만식', '이효석'],
'작품': ['봄봄', '동백꽃', '만무방',
'운수 좋은 날', 'B사감과 러브레터',
'날개', '권태',
'벙어리 삼룡이', '물레방아',
'레디메이드 인생', '메밀꽃 필 무렵'],
'연도': [1935, 1936, 1934,
1924, 1925,
1936, 1937,
1925, 1925,
1934, np.nan], # 이효석 '메밀꽃 필 무렵' 발표연도가 누락됨
'글자수': [12500, 9800, 13600,
11200, 7400,
18300, 14100,
16800, 14500,
22000, 11800],
})
authors = pd.DataFrame({
'작가': ['김유정', '현진건', '이상', '나도향', '채만식', '이효석', '염상섭'],
'생년': [1908, 1900, 1910, 1902, 1902, 1907, 1897],
'몰년': [1937, 1943, 1937, 1926, 1950, 1942, 1963],
})
```

```python
print(works)
print(authors)
```

```python
#Q1-a
print(works.shape)
print(works.dtypes)
works.describe()
#works, dtypes,describe의 속성을 프린트해야한다
```

```python
#짧은 관찰:NaN은 결측치를 나타내는 값으로, 일반적인 정수형 int에는 포함될 수 없다. 그래서 정수처럼 보이는 열에 NaN이 섞이면 pandas는 보통 그 열을 float64로 바꾸어 저장한다. 한편 숫자와 문자열 등 여러 자료형이 섞여 있으면 pandas는 더 포괄적인 자료형인 object로 처리한다.
```

```python
#Q1-b
print(works.isna().sum())
#isna().sum은 각 열에 대해 결측치의 개수를 알려준다
```

```python
#작가, 작품, 글자수 열에는 결측치가 없지만 연도 열에 결측치가 1개 있다
```

```python
#Q1-c
works2 = works.fillna(1936)
works2["연도"] = works2["연도"].astype(int)

print(works2.dtypes)
print(works2.tail(3))
#fillna()로 결측치를 채울 수 있고 astype()으로 형변환을 할 수 있다
```

```python
#Q2-1
after_1930=works2[works2["연도"]>=1930]["작품"]
print(after_1930)
kim_or_lee = works2.loc[works2["작가"].isin(["김유정", "이상"]), "작품"]
print(kim_or_lee)

#불인덱싱을 하여 연도가 1930년대 이후에 태어난 것들만 골라서 새로운 표를 만들 수 있고 loc을 이용해서 열 이름으로 접근할 수 있다
```

```python
#Q2-b
def categorize(n:int)->str:
    if n <10000:
        return "짧음"
    elif n >= 10000 and n <= 15000:
        return "보통"
    elif n>15000:
        return "긴"
works2["분량"]=works2["글자수"].apply(categorize)
works2[["작품","글자수","분량"]]

#미리 categorize라는 함수를 정의해두고 그것을 apply를 이용해서 적용해서 새로운 열을 추가할 수 있다
```

```python
print(works2.sort_values(["연도","글자수"],ascending=[True,False]))
#sort_values는 오름차순 혹은 내림차순으로 지정한 열을 정렬할 수 있다. 먼저 나온 열을 기준으로 정렬을 하고 같은 값일 경우 두번째 지정한 열을 기준으로 정렬한다
```

```python
#Q3-a
print(works2.groupby('작가')['글자수'].agg(["mean","count","sum"]))
#열의 값이 작가인 것들끼리 묶어서 글자수들의 평균, 총합, 횟수 등을 groupby와 agg로 고를 수 있다
```

```python
#Q3-b
print(works2["분량"].value_counts())
#value_counts는 열 이름이 분량에 해당하는 것에서 각 값들이 몇번 등장하는지 알려준다. 여기서 분량이라는 열에서 보통은 총 6번, 긴은 3번, 짧음은 2번 나온다
```

```python
#Q3-c
print(works2.groupby(["작가","분량"]).size())
#관찰: 가장 많은 양을 차지한 것은 김유정의 보통분량 소설이다. 이를 통해 김유정은 평소에 단편소설보다 중편 길이의 소설을 썼음을 확인할 수 있따
#그룹바이의 그룹키를 리스트로 주면 작가별로 한 번 묶고, 분량별로 한 번 묶는다
```

```python
#Q4-a
full=pd.merge(works2,authors,on="작가",how='left')
print(full)
print(full.shape)
#works2와 authors 데이터를 합치는데 작가라는 콜롬으로 합치고 왼쪽 데이터, 즉 works2를 기준으로 합친다
#확인질문:how= left즉 왼쪽에 있는 데이터인 works2를 기준으로 합쳐서 works2에 있는 것들만 포함하고 authors에만 있는 작가들은 포함하지 않는다
```

```python
# Q4-b
full["집필 당시 나이"] = full["연도"] - full["생년"]
print(full)
#새열을 추가하기 위해 full의 연도 콜롬을 가져오고 생년 콜롬을 가져와서 집필당시나이를 만든다
```

```python
#Q4-c
print(full.groupby("작가")['집필 당시 나이'].mean().sort_values(ascending=True))
#작가 콜롬에서 값이 같은 것 끼리 합치고 집필당시나이의 평균을 가져오고 그것을 오름차순으로 정리한다
#해석: 앞서 만든 full과 현재 출력된것으로 보았을 때 가장 어릴때 작품을 집필한 작가는 나도향을 23에 집필을 했고 평균 23살에 집필을 했다. 반면가장 늦은 나이에 집필을 한 것은 채만식이다
```
